# Answer Key
Below are the reference SQL solutions to the questions above.

### Question 1 Answer
**Task:** Identify the unique list of premium menu items that sell for over $10.00 each, sorted alphabetically.

```python
%%sql
SELECT DISTINCT
    item_name
FROM orders
WHERE item_price > 10.00
ORDER BY item_name ASC;
```

### Question 2 Answer
**Task:** Find our overall sales performance by calculating the total count of items sold and the gross revenue generated across all transactions.

```python
%%sql
SELECT
    SUM(quantity) AS total_items_sold,
    SUM(item_price) AS total_revenue
FROM orders;
```

### Question 3 Answer
**Task:** Evaluate performance by product. For each menu item, find its total volume sold and the aggregate sales revenue. Rank the items from our highest grossing to lowest.

```python
%%sql
SELECT
    item_name,
    SUM(quantity) AS total_quantity,
    SUM(item_price) AS total_revenue
FROM orders
GROUP BY item_name
ORDER BY total_revenue DESC;
```

### Question 4 Answer
**Task:** Segment our sales lines into pricing tiers. Categorize the unit price of each line item as 'Budget' (under `$5.00`), 'Standard' (`$5.00` to `$10.00` inclusive), or 'Premium' (over `$10.00`). List the product name, unit price, and price tier for the first 10 items alphabetically.

```python
%%sql
SELECT
    item_name,
    (item_price / quantity) AS unit_price,
    CASE
        WHEN (item_price / quantity) < 5.00 THEN 'Budget'
        WHEN (item_price / quantity) BETWEEN 5.00 AND 10.00 THEN 'Standard'
        ELSE 'Premium'
    END AS price_category
FROM orders
ORDER BY item_name ASC
LIMIT 10;
```

### Question 5 Answer
**Task:** Pinpoint our high-volume products. Find all items that have sold a total volume of at least 100 units across our history, ordered from our most popular products downwards.

```python
%%sql
SELECT
    item_name,
    SUM(quantity) AS total_quantity
FROM orders
GROUP BY item_name
HAVING SUM(quantity) >= 100
ORDER BY total_quantity DESC;
```

### Question 6 Answer
**Task:** Analyze custom orders featuring guacamole. Focus on dishes where the customer added guacamole, and find the total units sold alongside the average price paid. Only show dishes with a premium positioning where the average price exceeds $9.00, ordered from highest average price to lowest.

```python
%%sql
SELECT
    item_name,
    SUM(quantity) AS total_quantity,
    AVG(item_price) AS avg_price
FROM orders
WHERE choice_description LIKE '%Guacamole%'
GROUP BY item_name
HAVING AVG(item_price) > 9.00
ORDER BY avg_price DESC;
```

### Question 7 Answer
**Task:** Profile our burrito and taco products. For any item containing 'Burrito' or 'Taco' in its name, calculate the typical purchase size (average quantity per order line), as well as the cheapest and most expensive unit prices we have charged for them. Group the results by product name.

```python
%%sql
SELECT
    item_name,
    AVG(quantity) AS avg_quantity_per_line,
    MIN(item_price / quantity) AS min_unit_price,
    MAX(item_price / quantity) AS max_unit_price
FROM orders
WHERE item_name LIKE '%Burrito%' OR item_name LIKE '%Tacos%'
GROUP BY item_name
ORDER BY item_name ASC;
```

### Question 8 Answer
**Task:** Identify our largest or highest-value orders. Find all transactions where either the customer ordered more than 3 total items or the total transaction value exceeded $30.00. Return the order reference, total item count, and final transaction cost, ordered from our highest-spending customers down.

```python
%%sql
SELECT
    order_id,
    SUM(quantity) AS total_items_in_order,
    SUM(item_price) AS total_order_cost
FROM orders
GROUP BY order_id
HAVING SUM(quantity) > 3
    OR SUM(item_price) > 30.00
ORDER BY total_order_cost DESC;
```

### Question 9 Answer
**Task:** Analyze our salsa-customized orders. For each product where customers opted for salsa, determine the total volume sold, overall revenue, and the actual average price paid per unit. Only include products that brought in at least $200.00 in salsa-related sales, and display the top 5 highest-grossing products.

```python
%%sql
SELECT
    item_name,
    SUM(quantity) AS total_quantity,
    SUM(item_price) AS total_revenue,
    (SUM(item_price) / SUM(quantity)) AS weighted_avg_unit_price
FROM orders
WHERE choice_description LIKE '%Salsa%'
GROUP BY item_name
HAVING SUM(item_price) >= 200.00
ORDER BY total_revenue DESC
LIMIT 5;
```

### Question 10 Answer
**Task:** Compare the performance of our main product categories (Bowls, Burritos, Tacos, vs. Sides/Others). For each of these custom categories, evaluate the transaction volume (total order lines), total items sold, total revenue, and the average basket value per line. Exclude simple, non-customized orders where no choice description is present. Only display categories that generated over $1,000.00 in total sales, ranked by gross revenue.

```python
%%sql
SELECT
    CASE
        WHEN item_name LIKE '%Bowl%' THEN 'Bowl'
        WHEN item_name LIKE '%Burrito%' THEN 'Burrito'
        WHEN item_name LIKE '%Taco%' THEN 'Taco'
        ELSE 'Side/Other'
    END AS item_category,
    COUNT(*) AS total_order_lines,
    SUM(quantity) AS total_quantity_sold,
    SUM(item_price) AS total_revenue,
    AVG(item_price) AS avg_line_price
FROM orders
WHERE choice_description IS NOT NULL
GROUP BY
    CASE
        WHEN item_name LIKE '%Bowl%' THEN 'Bowl'
        WHEN item_name LIKE '%Burrito%' THEN 'Burrito'
        WHEN item_name LIKE '%Taco%' THEN 'Taco'
        ELSE 'Side/Other'
    END
HAVING SUM(item_price) > 1000.00
ORDER BY total_revenue DESC;
```
